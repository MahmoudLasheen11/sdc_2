## 3) Create a mini package and a CLI
We will create a simple package `mini_uv_project` with a module and a console entrypoint using `pyproject.toml` `[project.scripts]`.

Structure (will be created by the commands below):

- `app/`
  - `cli.py`

Entrypoint will be `mini-hello`.


````python
%%bash
mkdir -p app
````

````python
%%file app/__init__.py
__all__ = []
````

````python
%%file main.py
from __future__ import annotations
import argparse

def main(args: list[str] | None = None) -> int:
    parser = argparse.ArgumentParser(prog="mini-hello")
    parser.add_argument("name", nargs="?", default="world")
    ns, _unknown = parser.parse_known_args(args)  # <- tolerate extras like -q
    print(f"Hello, {ns.name} 👋 (from mini_uv_project)")
    return 0

if __name__ == "__main__":
    raise SystemExit(main())
````

You can either run the app by activating the environment and use

`python main.py`

or by using

`uv run main.py`

`uv run` -> executes any script in the setup project environment

````python
%%bash
source .venv/bin/activate
python main.py
````

````python
%%bash
uv run main.py Student
````

### About `pyproject.toml`

This file is the central configuration for the project, following [PEP 518](https://peps.python.org/pep-0518/). It defines:

- **[build-system]**  
  Tells tools how to build the project (here: using `setuptools` + `wheel`).

- **[project]**  
  Project metadata (name, version, Python requirement, dependencies).  
  Also declares console scripts so that `mini-hello` runs `main:main`.

- **[tool.setuptools]**  
  Configuration for setuptools about how to package the code.  
  For a single-file module, we use `py-modules = ["main"]`.

- **[dependency-groups]** *(optional)*  
  Dev dependencies like `pytest` and `ruff` for testing/linting.

This file replaces older configs like `setup.py` or `setup.cfg` and is the only file `uv` (and other modern build tools) need to install, build, and distribute the package.

Overwrite pyproject.toml with additional scripts instruction and setup tools

```
[project.scripts]
mini-hello = "main:main"

[tool.setuptools]
package-dir = { "" = "." }
py-modules = ["main"]  
```

````python
%%file pyproject.toml

[project]
name = "0-uv-tutorial" # 
version = "0.1.0"
description = "Add your description here"
readme = "README.md"
requires-python = ">=3.11"
dependencies = [
    "requests>=2.32.5",
]

[dependency-groups]
dev = [
    "pytest>=8.4.2",
    "ruff>=0.13.0",
]

[project.scripts]
mini-hello = "main:main" #<-- Add this section, points to main function in main.py

[tool.setuptools]
package-dir = { "" = "." } #<-- Add this section, points to the root directory
py-modules = ["main"]     #<-- Add this section, specifies main.py as module
````

Install the project and run CLI via CLI commands

## 4) Testing and linting
We'll add `pytest` for tests (already installed as dev dependency) and `ruff` for linting, and create a basic test to validate our CLI.

Commands:

```bash
uv add --dev ruff
uv run pytest -q
uv run ruff check .
```

We'll generate a minimal test file below and run both tools.


Creating a pytest configuration

````python
%%file pytest.ini
[pytest]
testpaths = tests
pythonpath = .
````

````python
%%bash
# Ensure tests directory exists
mkdir -p tests
````

````python
%%file tests/test_cli.py
from main import main

def test_main_runs(capsys):
    assert main() == 0
    captured = capsys.readouterr()
    assert "Hello, world" in captured.out
````

### Add ruff and run tests/lint via CLI

````python
%%bash
uv add --dev ruff
````

````python
%%bash
uv run pytest -q
uv run ruff check .
````
